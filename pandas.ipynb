```python
data_with_missing = {
    'cars': ['Bmw', 'Volvo', None, 'Tesla', 'Ferrari', None, "Ford"],
    'counts': [2, 5, None, 3, 1, 4, None]
}

df_missing = pd.DataFrame(data_with_missing)

print(df_missing)
```

```python
df_missing.isnull().sum()
```

```python
# dropna(datalardagi yoq malumotlarni ochirish methodlari )
df_missing.dropna() # argument sifatida (how="all"nibersak qiymati borini chiqarib beradi) subset=[keni belgilasa yoq malumotni olib tashlaydi]
```

```python
# fillna(malumoti yoq valuelarga berilga raqam yoki sozni qoyib qoyadi)
df_missing.fillna("yoq") # ({'cars': "yoq", 'counts': 0}) ikkinchi usul
```
