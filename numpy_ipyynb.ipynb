```python
import numpy as np
print('nump => python lybery')
print("ndarray_number, dimensional,")

print('1D_array-vector')
arr = np.array([1,3,5,7])

print("numpy_array:", arr)

arr.shape
```

```python
print('2D_array-matrix_matritsa')
arr2 = np.array([[1,3,5,7],[4,6,8,9]])

print('numpy_array:',arr2)

arr2.shape
```

```python
# dtype()
a = np.array([1,2,3])

print(a.dtype)
```

```python
# reshape()
a = np.array([1,2,3,4,5,6])
b = a.reshape(2,3)

print('shaped_array:',b)
```

```python
# astype()
a = np.array([1,2,3])
b = a.astype(float)

print("float:",b)
```

```python
# np.zeros(0-dan t-sh array), np.ones(1-dan t-sh array)=> dammy data
z = np.zeros((2,3))
o = np.ones((2,3))

print('zeros:',z)
print('ones:',o)
```

```python
# ndim(dimensional-aniqlash)
a = np.array([[1,2],[3,4]])

print('ndim:', a.ndim)
```

```python
# arange(ketmaketligini sanab beradi)
a = np.arange(1,6)
print('arange:',a)
```

```python
# sum(qoshadi), mean(ortachasi), max(eng yuqorisi), min(eng kichigini)
a = np.array([1,2,3,4])

print('sum:', a.sum())
print('mean:', a.mean())
print('max:', a.max())
print('min:', a.min())
```

```python
# flatten(2d 3d arrayni 1darrga aylantiradi)
a = np.array([[1,2],[3,4]])

print('flatten:', a.flatten())
```

```python
# T(ornini almashtirish)TRANSPOSE
a = np.array([[1,2],[3,4]])

print('TRANSPOSE:', a.T)
```

```python
# dot(kopaytirishda ishlatiladi) @
a = np.array([[2,3],[4,5]])
b = np.array([[2,3],[4,5]])

print('dot/kopaytirish:', a.dot(b))
print("a ning 1chi qatoriga b ning 1 ustuni a(2,3)b(2,4)2*2+3*4=16, a(2*3)b(3*5)6+15=21")
```

```python
# random.rand(xarxil raqamlarni yaratib beradi)
a = np.random.rand(2, 3) # => 0.0 and 1.0 / 0.0 included(oz ichiga oladi) / 1.0 excluded(dan tashqari)

print("random rand:", a)
```
